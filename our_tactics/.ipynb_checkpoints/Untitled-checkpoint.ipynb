```python
import xml.etree.ElementTree as ET

raw_path = REPO / "data" / "tactical" / "BAR-ATH_tactical_atd.txt"
meta_path = REPO / "data" / "tactical" / "BAR-ATH_tactical_atd.xml"

# Inspect raw tracking data
ids = []
print("FIRST 3 TRACKING ROWS:")
with open(raw_path, "r") as f:
    for i, line in enumerate(f):
        if i < 3:
            print(line[:400].rstrip())
        try:
            ids.append(int(line.split(":", 1)[0]))
        except ValueError:
            pass

print("\nTRACKING FRAME STATISTICS")
print("Rows with numeric frame IDs:", len(ids))
print("First 10 IDs:", ids[:10])
print("Last 10 IDs:", ids[-10:])
print("Minimum ID:", min(ids) if ids else None)
print("Maximum ID:", max(ids) if ids else None)
print("Unique IDs:", len(set(ids)))
print("Strictly increasing:", all(a < b for a, b in zip(ids, ids[1:])))

# Inspect XML metadata
root = ET.parse(meta_path).getroot()

print("\nMETADATA")
print("Root tag:", root.tag)
print("Root attributes:", root.attrib)

for elem in root.iter():
    tag = elem.tag.split("}")[-1]
    if tag.lower() in ("team", "player"):
        print(tag, elem.attrib, (elem.text or "").strip()[:100])
```
